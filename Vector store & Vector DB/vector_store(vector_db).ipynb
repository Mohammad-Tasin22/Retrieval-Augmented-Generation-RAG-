```python
from langchain_huggingface import HuggingFaceEmbeddings
```

```python
from langchain_community.vectorstores import Chroma
```

```python
from langchain_core.documents import Document
```

```python
doc1 = Document(
    page_content = "Generative AI is becoming increasingly popular in education. Students use AI tools to understand difficult topics, summarize chapters, generate practice questions, and improve their writing. Teachers can also use AI to prepare lesson plans, quizzes, and educational materials. One important advantage of AI is personalized learning. Students can ask the same question in different ways and receive explanations according to their level of understanding. For example, a beginner can ask an AI system to explain programming using simple examples, while an advanced learner can request a technical explanation.",
    metadata = {"topic" : "Generative AI in Education"}
)

doc2 = Document(
    page_content = "Generative AI is creating new possibilities in healthcare by helping professionals analyze information and improve medical workflows. AI systems can summarize medical documents, assist with clinical notes, generate patient-friendly explanations, and support medical research. Researchers are also exploring AI for drug discovery and the analysis of large amounts of scientific information.",
    metadata = {"topic" : "Generative AI in Healthcare"}
)

doc3 = Document(
    page_content = "However, excessive dependence on AI can create problems. Students may copy AI-generated answers without understanding the concepts. This can negatively affect critical thinking, creativity, and independent problem-solving skills. There are also concerns about inaccurate information and academic dishonesty.",
    metadata = {"topic" : "Generative AI in Education"}
)

doc4 = Document(
    page_content = "However, healthcare requires a high level of accuracy. AI-generated information may sometimes contain errors or misleading recommendations. Patient privacy is another major concern because medical information is highly sensitive. AI systems should therefore operate under strict security, privacy, and professional oversight.Generative AI is not expected to replace doctors. Instead, its most valuable role is supporting healthcare professionals with information processing and routine tasks while leaving important medical decisions to qualified humans.",
    metadata = {"topic" : "Generative AI in Healthcare"}

)
```

```python
docs = [doc1,doc2,doc3,doc4]
```

```python
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
vector_store = Chroma(
    embedding_function= embeddings,
    persist_directory = "my_chroma_db",
    collection_name = "sample"
)
```

```python
#add_documents
vector_store.add_documents(docs)
```

```python
#view documents
vector_store.get(include=['embeddings','documents','metadatas'])
```

```python
vector_store.similarity_search(
    query = "What are the risks of excessive AI dependence among students?",
    k = 3
)
```

```python
#meta data filtering

vector_store.similarity_search_with_score(
    query = "",
    filter = {"topic": "Generative AI in Education"}
)
```
