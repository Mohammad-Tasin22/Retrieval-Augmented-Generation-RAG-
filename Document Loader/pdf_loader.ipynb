```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_groq import ChatGroq
from langchain.chat_models import init_chat_model
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate
from dotenv import load_dotenv
import os
```

```python
loader = PyPDFLoader("CKD_Hypertension_Research_Proposal.pdf")
```

```python
document = loader.load()
```

```python
print(len(document))
```

```python
print(document[2].page_content)
```

```python
model = init_chat_model("openai/gpt-oss-120b", model_provider="groq")
```

```python
prompt = PromptTemplate(
    template = "What is the limitations discussed in this research proposal {proposal}?",
    input_variables = {"proposal"}
)
```

```python
parser = StrOutputParser()
```

```python
chain = prompt | model | parser
```

```python
print(chain.invoke({"proposal": document[2].page_content}))
```
